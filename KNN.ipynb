```python
from collections import Counter

# ==============================================================================
# 1. TẠO DỮ LIỆU GIẢ LẬP (3 Chiều: [Chiều cao (cm), Cân nặng (kg), Tuổi thọ])
# Nhãn: 'Poodle' (Chó nhỏ) hoặc 'Golden' (Chó to)
# ==============================================================================
train_data = [
    # Đặc trưng (Features)            # Nhãn (Label)
    ([25.0, 4.5, 14.0],              'Poodle'),
    ([28.0, 5.2, 15.0],              'Poodle'),
    ([24.0, 3.8, 13.5],              'Poodle'),
    ([30.0, 6.0, 14.2],              'Poodle'),
    ([58.0, 30.5, 11.0],             'Golden'),
    ([60.0, 32.0, 10.5],             'Golden'),
    ([55.0, 28.0, 12.0],             'Golden'),
    ([62.0, 34.0, 11.5],             'Golden'),
]

# ==============================================================================
# 2. THUẬT TOÁN K-NEAREST NEIGHBORS (FROM SCRATCH)
# ==============================================================================

def k_nearest_neighbors(data, query_point, k):
    """
    Tìm k điểm gần nhất trong tập dữ liệu so với query_point.
    Mỗi phần tử trong data là một tuple: (tọa_độ_n_chiều, nhãn).
    """
    all_points = []

    for features, label in data:
        # Tính khoảng cách Euclidean bình phương cho n chiều
        distance = sum((a - b) ** 2 for a, b in zip(features, query_point))
        all_points.append((distance, label, features))

    # Sắp xếp tăng dần theo khoảng cách
    all_points.sort(key=lambda item: item[0])

    # Lấy ra k điểm láng giềng gần nhất
    return all_points[:k]


def predict_knn(data, query_point, k=3):
    """
    Dự đoán nhãn cho điểm mới dựa trên cơ chế Majority Vote (bỏ phiếu số đông).
    """
    # Bước 1: Lấy k láng giềng gần nhất
    neighbors = k_nearest_neighbors(data, query_point, k)

    # Bước 2: Rút trích nhãn của k láng giềng
    neighbor_labels = [label for _, label, _ in neighbors]

    # Bước 3: Đếm số phiếu bầu cho từng nhãn
    vote_counts = Counter(neighbor_labels)
    predicted_label = vote_counts.most_common(1)[0][0]

    return predicted_label, neighbors


# ==============================================================================
# 3. CHẠY THỬ NGHIỆM TRÊN GOOGLE COLAB
# ==============================================================================
# Một chú chó lạ có: Cao 26 cm, Nặng 4.8 kg, 14 tuổi
test_dog = [26.0, 4.8, 14.0]
k_value = 3

prediction, neighbors = predict_knn(train_data, test_dog, k=k_value)

print(f"--- THÔNG TIN TRUY VẤN ---")
print(f"Đặc trưng chú chó cần đoán: {test_dog}")
print(f"Giá trị k = {k_value}\n")

print(f"--- KẾT QUẢ {k_value} LÁNG GIỀNG GẦN NHẤT ---")
for rank, (dist, label, feat) in enumerate(neighbors, 1):
    print(f"Top {rank}: Nhãn = {label:<7} | Đặc trưng = {feat} | Khoảng cách^2 = {dist:.2f}")

print(f"\n=> KẾT LUẬN DỰ ĐOÁN: Chú chó này là giống '{prediction}'")
```
